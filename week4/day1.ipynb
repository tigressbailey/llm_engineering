# The Chinchilla Scaling Law
Number of parameters: proportional to the number of training tokens

If you are getting diminishing returns from training with more training data, then this law gives you a rule of thumb for scaling your model.

And vice versa, if you upgrade to a model with double the number of weights, this law suggests you need twice the data.

## Chinchilla 缩放法则（中文笔记）

DeepMind 在 2022 年提出的经验规律：**模型参数量（weights）与训练数据量（tokens）应按比例一起增长，才能让算力用得最划算。** 经验比例约为 **每 1 个参数配约 20 个训练 token**。

### 三句话的含义
- **参数量 ∝ 训练 token 数**：模型大小和喂给它的数据量之间有一个大致固定的比例，不能只把模型做大却用很少的数据训练。
- **加数据收益递减时**：如果不停加数据但效果提升越来越小（diminishing returns），往往说明瓶颈在于模型太小，该增加参数量了。
- **反之亦然**：把参数量翻倍，就应把训练数据也翻倍，否则更大的模型会"吃不饱"。

### 生活化比喻
- **参数量** = 学生的脑容量（能记多少）
- **训练数据（token）** = 给学生看的书和资料

| 情况 | 结果 |
|------|------|
| 脑子很大，只给一本书 | 浪费脑力，学不到东西（模型大、数据少）|
| 脑子很小，给一图书馆的书 | 装不下，记不住（模型小、数据多）|
| 脑子大小 ↔ 书的数量 成比例 | 最高效，算力不浪费 ✅ |

### 为什么重要
Chinchilla 之前，GPT-3（1750 亿参数）属于"参数太多、数据太少"。DeepMind 用这条法则训练的 **Chinchilla 只有 700 亿参数**，但用了更多数据，性能反而**超过了 GPT-3**。

> 核心结论：固定算力预算下，与其盲目把模型做大，不如让"参数量"和"数据量"保持平衡增长，这样最划算、效果最好。

## 6 个高阶基准测试（中文笔记）

这些是用来**评估大模型能力上限**的高难度测试集。简单测试现在的模型早就刷满分了，所以用这些更难的来区分"谁更聪明"。

| 基准 | 考什么 | 一句话解释 |
|------|--------|-----------|
| **HLE**（Humanity's Last Exam，人类最后的考试）| 超人智能 | 各领域专家出题，难到连人类专家都未必会做 |
| **GPQA**（Graduate-level Google-Proof Q&A）| 博士级科学 | 物理/化学/生物难题，"Google 也搜不到"，必须真懂 |
| **MMLU-PRO** | 综合知识理解 | 经典 MMLU 加强版，几十个学科的硬核问答 |
| **AIME**（American Invitational Mathematics Examination）| 数学 | 美国高中顶级数学竞赛题，考数学推理 |
| **LiveCodeBench** | 编程 | 持续收集最新编程题，防止模型"背答案" |
| **MuSR**（Multistep Soft Reasoning）| 推理 | 多步逻辑推导（如谜题、情景分析）|

> 简单说：这 6 个测试分别从**通用智能、科学、综合知识、数学、编程、推理**六个角度，衡量一个大模型到底有多强。


## Limitations of Benchmarks when comparing LLMs
- Training Data Contamination
- Not always consistently applied
- Too narrow in scope
- Hard to measure nuanced reasoning
- Saturation
- Overfitting
- To be confirmed: Frontier LLMs many be aware that they are being evaluated on these benchmarks

## 基准测试的局限性（中文笔记）

核心观点：**用 benchmark 分数判断"哪个模型更强"有很多坑，分数高不一定真的强。**

| 局限 | 含义 | 通俗理解 |
|------|------|---------|
| **Training Data Contamination**（训练数据污染）| 测试题答案可能已被爬进训练数据，模型提前"见过" | 提前看过考卷 |
| **Not always consistently applied**（测试不统一）| 各团队的提示词、参数、评测方式不同，分数没法公平对比 | 各家考试规则不同，没法比 |
| **Too narrow in scope**（范围太窄）| 每个测试只考某一方面，满分不代表真实任务也好用 | 只考一科，不代表全能 |
| **Hard to measure nuanced reasoning**（难测细微推理）| 标准答案题测不出判断力、创造力等"软实力" | 软实力测不出来 |
| **Saturation**（饱和）| 测试太简单被刷爆，顶尖模型都接近满分，失去区分度 | 题太简单，都满分了 |
| **Overfitting**（过拟合）| 专门针对榜单优化，只擅长应付特定测试，真实任务下降 | 只为应试而学 |
| **可能"知道"在被测**（待确认）| 顶尖模型也许能识别"正在被评估"，从而表现得和平时不一样 | 考试时刻意装好学生 |

> 结论：benchmark 分数只能作参考，不能完全代表模型在真实使用中的好坏。
